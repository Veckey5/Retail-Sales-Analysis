**QUERIES - REVENUE AND PERFORMANCE ANALYSIS**

I answered six everyday requests, the kind a manager or teammate will need to know.

```sql
USE CATALOG practice; USE SCHEMA retail;
```

```sql

--1. Which premium products over $100 do we sell? Sorted from most to least expensive.
SELECT product_id, product_name, price
FROM products
WHERE price > 100
ORDER BY price DESC;
```

```sql
--2. Provide the regional manager with a contact list for the TX and CO markets.
SELECT customer_id, CONCAT(first_name, ' ', last_name) AS full_name, email, city, state
FROM customers
WHERE state IN ('TX', 'CO');
 
```

```sql
--3. Which products are out of stock right now? (with stock_quantity = 0)
SELECT product_id, product_name, stock_quantity
FROM products
WHERE stock_quantity = 0;
```

```sql
--4. Provide every order placed in March 2024
SELECT order_id, customer_id, order_date, status
FROM orders
WHERE order_date >= '2024-03-01' AND order_date < '2024-04-01';
```

```sql
--5. Give the Electronics team a readable product list - all products in the "Electronics" category
SELECT p.product_id, p.product_name, c.category_name, p.price
FROM products p
JOIN categories c ON p.category_id = c.category_id
WHERE c.category_name = 'Electronics';
```

```sql
--6. For every order placed, provide full details showing the customer's full name, the order date, and the order status.
SELECT o.order_id,
       CONCAT(c.first_name, ' ', c.last_name) AS full_name,
       o.order_date,
       o.status
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
ORDER BY o.order_date;
```
