**DATA VALIDATION**

In this stage of my analysis, I checked the data before analyzing it. This includes row counts, duplicates and price inconsistencies.

```sql
USE CATALOG practice; USE SCHEMA retail;
```

```sql
-- 1a. Row counts per table
SELECT 'categories'  AS table_name, COUNT(*) AS row_count FROM categories
UNION ALL SELECT 'customers',   COUNT(*) FROM customers
UNION ALL SELECT 'products',    COUNT(*) FROM products
UNION ALL SELECT 'orders',      COUNT(*) FROM orders
UNION ALL SELECT 'order_items', COUNT(*) FROM order_items;
```

```sql
-- 1b. Orphaned foreign keys (every query below should return 0 rows)
SELECT o.order_id, o.customer_id
FROM orders o
LEFT JOIN customers c ON o.customer_id = c.customer_id
WHERE c.customer_id IS NULL;
 
SELECT oi.order_item_id, oi.order_id
FROM order_items oi
LEFT JOIN orders o ON oi.order_id = o.order_id
WHERE o.order_id IS NULL;
 
SELECT oi.order_item_id, oi.product_id
FROM order_items oi
LEFT JOIN products p ON oi.product_id = p.product_id
WHERE p.product_id IS NULL;
```

```sql
-- 1c. Price consistency: does the price charged match the catalog price? If a mismatch exists, then there's a discount or a price change that should be investigated.
SELECT oi.order_item_id, oi.unit_price AS charged_price, p.price AS catalog_price
FROM order_items oi
JOIN products p ON oi.product_id = p.product_id
WHERE oi.unit_price <> p.price;
```

```sql
-- 1d. Duplicate primary keys (should return 0 rows)
SELECT order_id, COUNT(*) AS dupes
FROM orders
GROUP BY order_id
HAVING COUNT(*) > 1;
 
```

```sql
-- 1e. Order status values to confirm the three expected values: completed, returned, and cancelled.
SELECT DISTINCT status FROM orders;
```
