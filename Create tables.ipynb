**TABLE CREATION**

I built a small retail database from scratch that mimics a real online store (customers, products, orders) before writing any analysis. Then, I created a catalog and schema first so all my tables live in one tidy place instead of mixing with other work. 

**Note**
1. USE tells Databricks to build everything inside it.
2. I used CREATE OR REPLACE so I can re-run the whole script without "table already exists" errors.

```sql
CREATE CATALOG IF NOT EXISTS practice;
CREATE SCHEMA  IF NOT EXISTS practice.retail;
USE CATALOG practice;
USE SCHEMA retail;
```
